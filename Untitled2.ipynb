Print Format Method

```python
num1,num2,num3=10,20,30

add=num1+num2+num3

print('the addition of {} and {} and {} is={}'.format(num1,num2,num3,add))
# number of variables == no. of flower braces
```

```python
num1,num2,num3=10,20,30

add=num1+num2+num3

#format string var= {}
print('the addition of {} and {} and {} is={}'.format(num1,num2,num3,add))
#f string var within braces
print(f'the addition of {num1} and {num2} and {num3} is= {add}')
# number of variables == no. of flower braces
```
